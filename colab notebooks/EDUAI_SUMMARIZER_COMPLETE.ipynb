```python
from datasets import load_dataset, concatenate_datasets

# Loading datasets
print('Loading arxiv dataset...')
arxiv = load_dataset('ccdv/arxiv-summarization', trust_remote_code=True)
arxiv_train = arxiv['train'].select(range(5000))
arxiv_val = arxiv['validation'].select(range(500))

print('Loading CNN/DailyMail dataset...')
cnn = load_dataset('cnn_dailymail', '3.0.0')
cnn_train = cnn['train'].select(range(5000))
cnn_val = cnn['validation'].select(range(500))

print(f'Arxiv train: {len(arxiv_train)}, val: {len(arxiv_val)}')
print(f'CNN train: {len(cnn_train)}, val: {len(cnn_val)}')
```

```python
def format_arxiv(example):
    return {
        'input_text': 'Summarize the following academic text:\n' + example['article'][:2000],
        'target_text': example['abstract'][:500]
    }

def format_cnn(example):
    return {
        'input_text': 'Summarize the following text:\n' + example['article'][:2000],
        'target_text': example['highlights'][:500]
    }

arxiv_train = arxiv_train.map(format_arxiv, remove_columns=arxiv_train.column_names)
arxiv_val = arxiv_val.map(format_arxiv, remove_columns=arxiv_val.column_names)
cnn_train = cnn_train.map(format_cnn, remove_columns=cnn_train.column_names)
cnn_val = cnn_val.map(format_cnn, remove_columns=cnn_val.column_names)

train_dataset = concatenate_datasets([arxiv_train, cnn_train]).shuffle(seed=42)
val_dataset = concatenate_datasets([arxiv_val, cnn_val]).shuffle(seed=42)

print(f'Combined train: {len(train_dataset)}, val: {len(val_dataset)}')
print(f'\nSample input (first 200 chars): {train_dataset[0]["input_text"][:200]}')
print(f'\nSample target (first 200 chars): {train_dataset[0]["target_text"][:200]}')
```

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

MODEL_NAME = 'google/flan-t5-base'

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

print(f'Model loaded: {MODEL_NAME}')
print(f'Parameters: {sum(p.numel() for p in model.parameters()):,}')
```

```python
MAX_INPUT_LENGTH = 512
MAX_TARGET_LENGTH = 200

def tokenize_function(example):
    model_inputs = tokenizer(
        example['input_text'],
        max_length=MAX_INPUT_LENGTH,
        truncation=True,
        padding='max_length'
    )

    labels = tokenizer(
        example['target_text'],
        max_length=MAX_TARGET_LENGTH,
        truncation=True,
        padding='max_length'
    )

    labels_ids = [
        token if token != tokenizer.pad_token_id else -100
        for token in labels['input_ids']
    ]

    model_inputs['labels'] = labels_ids
    return model_inputs

print('Tokenizing train dataset...')
tokenized_train = train_dataset.map(tokenize_function, batched=False, remove_columns=train_dataset.column_names)

print('Tokenizing validation dataset...')
tokenized_val = val_dataset.map(tokenize_function, batched=False, remove_columns=val_dataset.column_names)

print(f'Tokenized train: {len(tokenized_train)}, val: {len(tokenized_val)}')
```

```python
from transformers import Seq2SeqTrainer, Seq2SeqTrainingArguments, DataCollatorForSeq2Seq

training_args = Seq2SeqTrainingArguments(
    output_dir=SAVE_DIR,

    num_train_epochs=3,
    per_device_train_batch_size=4,
    per_device_eval_batch_size=4,
    learning_rate=3e-5,
    weight_decay=0.01,
    warmup_steps=200,
    max_grad_norm=1.0,

    eval_strategy='epoch',
    predict_with_generate=True,
    generation_max_length=MAX_TARGET_LENGTH,

    save_strategy='epoch',
    save_total_limit=3,
    load_best_model_at_end=True,
    metric_for_best_model='eval_loss',

    fp16=False,
    bf16=False,

    logging_steps=50,
    report_to='none',
)

data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train,
    eval_dataset=tokenized_val,
    data_collator=data_collator,
)
```

```python
trainer.train()
```

```python
import glob
checkpoints = sorted(glob.glob(f'{SAVE_DIR}/checkpoint-*'))
if checkpoints:
  latest = checkpoints[-1]
  print(f'Resuming from: {latest}')
  trainer.train(resume_from_checkpoint=latest)
else:
  print('No checkpoints found.')
```

```python
import evaluate
import numpy as np
from tqdm import tqdm

rouge = evaluate.load('rouge')

print('Generating summaries for ROUGE evaluation...')
predictions = []
references = []

eval_samples = val_dataset.select(range(min(200, len(val_dataset))))

model.eval()
import torch
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

for i in tqdm(range(len(eval_samples))):
    input_text = eval_samples[i]['input_text']
    reference = eval_samples[i]['target_text']

    inputs = tokenizer(input_text, max_length=MAX_INPUT_LENGTH, truncation=True, return_tensors='pt').to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_length=MAX_TARGET_LENGTH,
            min_length=30,
            num_beams=4,
            length_penalty=1.0,
            no_repeat_ngram_size=3,
            early_stopping=True
        )

    prediction = tokenizer.decode(outputs[0], skip_special_tokens=True)
    predictions.append(prediction)
    references.append(reference)

# ROUGE
results = rouge.compute(predictions=predictions, references=references)

print('\n=========================================')
print('      ROUGE EVALUATION RESULTS')
print('=========================================')
print(f'ROUGE-1: {results["rouge1"]:.4f}')
print(f'ROUGE-2: {results["rouge2"]:.4f}')
print(f'ROUGE-L: {results["rougeL"]:.4f}')
print(f'ROUGE-Lsum: {results["rougeLsum"]:.4f}')
print('=========================================')

# Save results
with open(f'{SAVE_DIR}/rouge_results.txt', 'w') as f:
    for k, v in results.items():
        f.write(f'{k}: {v:.4f}\n')
print(f'Results saved to {SAVE_DIR}/rouge_results.txt')
```

```python
!pip install pdfplumber -q
import pdfplumber
import re

def extract_text_from_pdf(file_path):
    full_text = ''
    with pdfplumber.open(file_path) as pdf:
        for page in pdf.pages:
            text = page.extract_text()
            if text:
                full_text += text + '\n'
    return full_text

def clean_text(text):
    text = re.sub(r'\n+', '\n', text)
    text = re.sub(r'\s+', ' ', text)
    return text.strip()

from google.colab import files
print('Upload a PDF file to test summarization:')
uploaded = files.upload()

file_name = list(uploaded.keys())[0]
raw_text = extract_text_from_pdf(file_name)
cleaned = clean_text(raw_text)

print(f'\nExtracted {len(cleaned)} characters from {file_name}')
print(f'First 500 chars: {cleaned[:500]}')
```

```python
def summarize_long_text(text, chunk_size=1500, max_summary_length=200):
    words = text.split()
    chunks = []
    for i in range(0, len(words), chunk_size):
        chunk = ' '.join(words[i:i + chunk_size])
        chunks.append(chunk)

    print(f'Document split into {len(chunks)} chunks')

    chunk_summaries = []
    for i, chunk in enumerate(chunks):
        input_text = 'Summarize the following academic text:\n' + chunk
        inputs = tokenizer(input_text, max_length=MAX_INPUT_LENGTH, truncation=True, return_tensors='pt').to(device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_length=max_summary_length,
                min_length=30,
                num_beams=4,
                length_penalty=1.0,
                no_repeat_ngram_size=3,
                early_stopping=True
            )

        summary = tokenizer.decode(outputs[0], skip_special_tokens=True)
        chunk_summaries.append(summary)
        print(f'  Chunk {i+1}/{len(chunks)} summarized')

    if len(chunk_summaries) > 1:
        combined = ' '.join(chunk_summaries)
        input_text = 'Summarize the following text into a coherent summary:\n' + combined
        inputs = tokenizer(input_text, max_length=MAX_INPUT_LENGTH, truncation=True, return_tensors='pt').to(device)

        with torch.no_grad():
            outputs = model.generate(
                **inputs,
                max_length=300,
                min_length=50,
                num_beams=4,
                length_penalty=1.0,
                no_repeat_ngram_size=3,
                early_stopping=True
            )

        final_summary = tokenizer.decode(outputs[0], skip_special_tokens=True)
    else:
        final_summary = chunk_summaries[0]

    return final_summary, chunk_summaries

final_summary, chunk_sums = summarize_long_text(cleaned)



if len(chunk_sums) > 1:
    print(f'\n--- Individual Chunk Summaries ({len(chunk_sums)} chunks) ---')
    for i, cs in enumerate(chunk_sums):
        print(f'\nChunk {i+1}: {cs}')
```
